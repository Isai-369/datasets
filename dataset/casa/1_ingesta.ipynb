```python
import pandas as pd
import glob
```

```python
lista_csv = glob.glob('casa_*.csv')
lista_csv.sort()
```

```python
df = pd.concat( (pd.read_csv(f) for f in lista_csv), ignore_index=True)
```

```python
df.drop(columns=['RESPNO'],inplace=True)
```

```python
df.columns = df.columns.astype(str).str.lower()
```

```python
df =df [ ['lotsize', 'bed', 'bath', 'rooms', 'age', 'taxes', 'garage',
       'locate', 'type', 'fuel', 'heat', 'pool', 'eik', 'cac', 'firepl',
       'sewer', 'base', 'modkit', 'modbath','value'] ]
```

```python
df.to_csv('casas.csv',index=False)
```
